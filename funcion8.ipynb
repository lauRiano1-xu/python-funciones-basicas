# Función zip():
Permite combinar dos o más colecciones, agrupando sus elementos según la posición que ocupan.
## Sintaxis: 
zip (iterable1, iterable2)

```python
productos = ["Mouse", "Teclado", "Monitor"]
precios = [45000, 80000, 650000]

inventario = zip(productos, precios)

for producto, precio in inventario:
    print(producto, ":", precio)
```

## Explicación: 
Combina cada producto con su respectivo precio, facilitando la organización de información relacionada.
