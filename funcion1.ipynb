# Función len():
Permite conocer la cantidad de elementos que contiene un objeto. Puede utilizarse con cadenas de texto, listas, tuplas, conjuntos y diccionarios.
## Sintaxis: 
len(objeto) 

```python
productos = ["Mouse", "Teclado", "Monitos", "Audifonos"]
cantidad = len(productos)
print("cantidad de productos", cantidad) 
```

## Explicación:
La función cuenta los elementos almacenados en la lista y devuelve el número total.
