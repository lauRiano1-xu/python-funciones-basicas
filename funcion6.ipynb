# Función dict()
Permite crear un diccionario, que es una estructura de datos que almacena información mediante pares de clave y valor.
## Sintaxis: 
dict(clave=valor)

```python
estudiante = dict(
    nombre="Valentina Riaño",
    carrera="Derecho",
    semestre=8
)

print(estudiante)
print("Nombre:", estudiante["nombre"])
print("Carrera:", estudiante["carrera"])
```

# Explicación:
Se crea un diccionario para almacenar los datos de un estudiante y acceder a ellos mediante sus respectivas claves.
