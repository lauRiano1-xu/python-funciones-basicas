# Función tuple()
Permite crear una tupla a partir de un objeto iterable. Las tuplas son estructuras de datos que permiten almacenar varios elementos y a diferencia de las listas, no pueden modificarse después de su creación.
## Sintaxis: 
tuple (iterable)

```python
materias = ["Matematicas", "Programación", "Bases de daos"]
materias_fijas = tuple(materias)
print(materias_fijas)
```

## Explicación: 
Convierte la lista de materias en una tupla, conservando sus elementos.
