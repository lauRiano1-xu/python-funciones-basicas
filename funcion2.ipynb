# Función sorted():
Permite ordenar los elementos de una colección y devuelve una nueva lisata con los elementos organizados. Por defecto, los ordena de menor a mayor o alfabéticamente.
## Sintaxis: 
sorted(iterable, reverse=false) 

```python
nombres = ["Sebastian", "Brandon", "Valentina", "Laura "]
nombres_ordenados = sorted(nombres)
print("Nombres originales:", nombres)
print("nombres ordenados:", nombres_ordenados)
```

## Explicación: 
La función organiza los nombres alfabéticamente sin modificar la lista original
