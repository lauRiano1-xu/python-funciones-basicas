# Función csv.reader():
La función reader() del módulo csv permite leer información almacenada en archivos CSV, organizando los datos en filas y columnas.
## Sintaxis:
csv.reader(archivo)
## Ejemplo:
Leer un archivo CSV que contiene nombres y edades de diferentes personas.

```python
import csv

with open("personas.csv", "w", newline="", encoding="utf-8") as archivo:
    archivo.write("nombre,edad\nLaura,21\nPedro,24\n")

with open("personas.csv", "r", newline="", encoding="utf-8") as archivo:
    lector = csv.reader(archivo)

    for fila in lector:
        print(fila)
```

## Explicación:
La función csv.reader() interpreta el contenido del archivo CSV y permite recorrer cada fila. Cada registro se devuelve como una lista de valores.
