# Función input():
permite solicitar información al usuario mientras se ejecuta un programa. Los datos ingresados se reciben como cadenas de texto (str).
## Sintaxis:
input("Mensaje para el usuario")
## Ejemplo:
Solicitar el nombre y la edad de una persona para mostrar sus datos.

```python
nombre = input("Ingresa tu nombre: ")
edad = input("Ingresa tu edad: ")

print("Hola", nombre)
print("Tu edad es:", edad, "años")
```

## Explicación:
El programa solicita dos datos al usuario y posteriormente los muestra en pantalla. La función input() recibe la información escrita y la almacena en las variables correspondientes.
