# Método read():
permite obtener el contenido de un archivo que previamente ha sido abierto. Devuelve la información como una cadena de texto.
## Sintaxis:
archivo.read()
## Ejemplo:
Leer los productos almacenados en un archivo de texto.

```python
with open("productos.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Teclado\nMouse\nMonitor")

with open("productos.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
```

## Explicación:
El método read() lee todo el contenido del archivo y lo almacena en la variable contenido. Finalmente, print() muestra los productos registrados.
