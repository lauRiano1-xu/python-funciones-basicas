# Método sys.stdin.readline():
permite leer una sola línea de información desde la entrada estándar del programa, como los datos recibidos desde la terminal.
## Sintaxis:
sys.stdin.readline()
## Ejemplo:
Leer el nombre de una ciudad desde la entrada estándar.

```python
import sys
from io import StringIO

entrada_original = sys.stdin

sys.stdin = StringIO("Bogotá\n")

ciudad = sys.stdin.readline().strip()

sys.stdin = entrada_original

print("Ciudad ingresada:", ciudad)
```

## Explicación: 
El método readline() obtiene una línea de información. En este ejemplo, StringIO simula la entrada de datos y strip() elimina el salto de línea que queda al final.
