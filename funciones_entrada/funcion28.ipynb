# Método readlines():
permite leer todas las líneas de un archivo y almacenarlas en una lista de Python.
## Sintaxis:
archivo.readlines()
## Ejemplo:
Leer una lista de nombres almacenados en un archivo y mostrarlos como una colección.

```python
with open("nombres.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Sofía\nAndrés\nValentina\n")

with open("nombres.txt", "r", encoding="utf-8") as archivo:
    nombres = archivo.readlines()

nombres = [nombre.strip() for nombre in nombres]

print(nombres)
```

## Explicación:
El método readlines() obtiene todas las líneas del archivo y las guarda en una lista. Posteriormente, strip() elimina los saltos de línea de cada elemento.
