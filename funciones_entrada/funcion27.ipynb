# Método readline():
permite leer una línea específica de un archivo de texto. Es útil cuando se necesita consultar información de manera individual.
## Sintaxis:
archivo.readline()
## Ejemplo:
Consultar el primer estudiante registrado en un archivo.

```python
with open("estudiantes.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Ana\nCarlos\nMaría")

with open("estudiantes.txt", "r", encoding="utf-8") as archivo:
    primera_linea = archivo.readline().strip()

print("Primer estudiante:", primera_linea)
```

## Explicación:
El método readline() obtiene únicamente la primera línea del archivo. Se utiliza strip() para eliminar el salto de línea y mostrar el nombre correctamente.
