# Función getpass():
permite solicitar información confidencial, como una contraseña, sin mostrar los caracteres que el usuario está escribiendo en la consola.
## Sintaxis:
getpass("Mensaje para el usuario")
## Ejemplo:
Solicitar una contraseña y mostrar la cantidad de caracteres ingresados.

```python
from getpass import getpass

contrasena = getpass("Ingresa tu contraseña: ")

print("La contraseña tiene", len(contrasena), "caracteres.")
```

## Explicación:
La función getpass() recibe la contraseña sin mostrarla mientras se escribe. Después, len() permite conocer la cantidad de caracteres que contiene.
