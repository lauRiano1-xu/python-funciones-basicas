# Función json.load():
permite leer información almacenada en archivos JSON y convertirla en estructuras de datos de Python, como diccionarios y listas.
## Sintaxis: 
json.load(archivo)
## Ejemplo:
Consultar la información de un producto almacenado en un archivo JSON.

```python
import json

datos = {
    "producto": "Teclado",
    "precio": 85000,
    "disponible": True
}

with open("producto.json", "w", encoding="utf-8") as archivo:
    json.dump(datos, archivo, ensure_ascii=False)

with open("producto.json", "r", encoding="utf-8") as archivo:
    producto = json.load(archivo)

print("Producto:", producto["producto"])
print("Precio:", producto["precio"])
print("Disponible:", producto["disponible"])
```

## Explicación:
La función json.load() lee los datos del archivo JSON y los convierte en un diccionario de Python. De esta manera, podemos consultar cada elemento utilizando su respectiva clave.
