# Función open():
permite abrir archivos para leer, escribir o modificar información almacenada en ellos.
## Sintaxis:
open("nombre_archivo", "modo")
## Ejemplo:
Crear un archivo de texto, escribir un mensaje y posteriormente abrirlo para consultar su contenido.

```python
with open("mensaje.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Bienvenidos al curso de Python.")

with open("mensaje.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
```

## Explicación:
La función open() permite trabajar con archivos. Primero se utiliza el modo "w" para escribir información y después el modo "r" para leerla. La instrucción with permite cerrar automáticamente el archivo al terminar.
