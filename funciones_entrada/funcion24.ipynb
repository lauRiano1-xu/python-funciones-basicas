# Método sys.stdin.read():
permite obtener todo el contenido disponible desde la entrada estándar, incluyendo varias líneas de información.
## Sintaxis:
sys.stdin.read()
## Ejemplo:
Leer varias líneas de texto y mostrarlas en pantalla.

```python
import sys
from io import StringIO

entrada_original = sys.stdin

sys.stdin = StringIO("Primera línea\nSegunda línea\n")

texto = sys.stdin.read()

sys.stdin = entrada_original

print(texto, end="")
```

## Explicación:
El método read() captura todo el contenido disponible en la entrada estándar, sin limitarse a una sola línea. En este ejercicio se utiliza StringIO para simular los datos recibidos.
