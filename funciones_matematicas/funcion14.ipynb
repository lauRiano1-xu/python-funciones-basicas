# Función divmode()
permite realizar una división y obtener simultáneamente el cociente y el residuo. Devuelve ambos resultados en una tupla.
## Sintaxis: 
divmod(dividendo, divisor)
## Ejemplo:
Una empresa tiene  50 productos que necesita organizar en cajas con capacidad para 6 productos cada una. Puede utilizar divmod() para conocer cuántas cajas completas puede llenar y cuántos productos sobran.

```python
productos = 50
capacidad = 6

resultado = divmod(productos, capacidad)

print("Cajas completas:", resultado[0])
print("Productos sobrantes:", resultado[1])
```

## Explicación:
La función devuelve una tupla (8, 2): se pueden llenar 8 cajas completas y quedan 2 productos sin empacar.
