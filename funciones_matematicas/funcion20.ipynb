# Función complex():
permite crear números complejos, los cuales están compuestos por una parte real y una parte imaginaria. En Python, la parte imaginaria se representa mediante la letra j.
## Sintaxis:
complex(parte_real, parte_imaginaria)
## Ejemplo:
En un ejercicio de matemáticas se necesita representar un número complejo cuya parte real es 5 y cuya parte imaginaria es 3.

```python
numero = complex(5, 3)

print("Número complejo:", numero)
print("Parte real:", numero.real)
print("Parte imaginaria:", numero.imag)
```

## Explicación:
La función complex() crea el número complejo (5+3j), donde 5 representa la parte real y 3 representa la parte imaginaria.
