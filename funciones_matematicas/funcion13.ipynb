# Función pow()
permite elevar un número a una potencia determinada. Recibe una base y un exponente, y devuelve el resultado de la operación.
## Sintaxis:
pow(base, exponente)
## Ejemplo:
Un estudiante necesita calcular el área de un terreno cuadrado que tiene 8 metros por cada lado. Para ello, puede elevar el valor 8 al cuadrado.

```python
lado = 8

area = pow(lado, 2)

print("Medida del lado:", lado, "metros")
print("Área del terreno:", area, "metros cuadrados")
```

## Explicación:
La función pow() eleva 8 a la potencia 2, realizando la operación 8 × 8, cuyo resultado es 64.
