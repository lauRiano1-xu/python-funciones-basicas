# Función abs():
permite obtener el valor absoluto de un número. Esto significa que devuelve su valor sin importar si es positivo o negativo. Es útil para calcular distancias, diferencias y variaciones numéricas.
## Sintaxis:
abs(numero)
## Ejemplo:
Imagina que una persona tiene una deuda de $85.000. Para conocer el valor de esa cantidad sin tener en cuenta el signo negativo, podemos utilizar abs().

```python
deuda = -85000

valor_absoluto = abs(deuda)

print("Valor original:", deuda)
print("Valor absoluto:", valor_absoluto)
```

## Explicación del resultado:
La función abs() recibe el valor -85000 y devuelve 85000, eliminando el signo negativo sin modificar la variable original.
