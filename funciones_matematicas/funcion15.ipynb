# Función sum()
permite sumar todos los elementos numéricos de una colección, como una lista o una tupla.
## Sintaxis: 
sum(iterable)
## Ejemplo:
Un estudiante desea conocer el total de sus gastos diarios durante cinco días. Puede almacenar los valores en una lista y utilizar sum() para calcular el total.

```python
gastos = [12000, 8500, 15000, 10000, 9500]

total_gastos = sum(gastos)

print("Gastos registrados:", gastos)
print("Total de gastos:", total_gastos)
```

## Explicación:
La función sum() suma todos los valores de la lista y devuelve un total de 55.000.
