# Función int():
permite convertir un valor numérico o una cadena que representa un número entero en un dato de tipo entero. Cuando recibe un número decimal, elimina su parte fraccionaria sin redondear.
## Sintaxis:
int(valor)
## Ejemplo:
Un sistema recibe la cantidad de productos como un número decimal, pero necesita trabajar únicamente con unidades enteras.

```python
cantidad = 15.8

cantidad_entera = int(cantidad)

print("Cantidad original:", cantidad)
print("Cantidad convertida:", cantidad_entera)
print("Tipo de dato:", type(cantidad_entera))
```

## Explicación:
La función int() convierte el valor decimal en un entero, eliminando los decimales. Es importante recordar que no realiza un redondeo.
