# Función round()
permite redondear un número decimal al entero más cercano o a una cantidad determinada de posiciones decimales.
## Sintaxis:
round(numero, cantidad_decimales) 
## Ejemplo: 
Una tienda calcula el precio de un producto y obtiene un valor decimal de $15999.678. Para presentar un precio con dos decimales, puede utilizar round().

```python
precio = 15999.678

precio_redondeado = round(precio, 2)

print("Precio original:", precio)
print("Precio redondeado:", precio_redondeado)
```

## Explicación:
La función round() aproxima el número a dos posiciones decimales, facilitando la presentación de valores numéricos.
