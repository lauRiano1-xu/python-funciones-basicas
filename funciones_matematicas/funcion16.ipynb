# Función min()
permite identificar y devolver el valor más pequeño dentro de una colección de elementos o entre varios valores proporcionados.
## sintaxis:
min(iterable)
## Ejemplo:
Un profesor necesita identificar la calificación más baja de un grupo de estudiantes. Para hacerlo, puede utilizar min() sobre una lista de notas.

```python
calificaciones = [4.5, 3.8, 2.9, 4.2, 3.5]

nota_menor = min(calificaciones)

print("Calificaciones:", calificaciones)
print("Nota más baja:", nota_menor)
```

## Explicación:
La función min() compara los valores de la lista y devuelve el menor, que en este caso es 2.9.
