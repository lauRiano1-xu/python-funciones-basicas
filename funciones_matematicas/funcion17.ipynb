# Función max()
permite encontrar y devolver el valor más grande dentro de una colección de elementos o entre varios valores.
## Sintaxis:
max(iterable)
## Ejemplo:
Una tienda desea identificar cuál fue el producto que tuvo el precio más alto dentro de una lista de precios registrados.

```python
precios = [25000, 45000, 32000, 78000, 56000]

precio_mayor = max(precios)

print("Precios registrados:", precios)
print("Precio más alto:", precio_mayor)
```

## Explicación:
La función max() compara los valores y devuelve el número más grande de la lista, que corresponde a 78.000.
