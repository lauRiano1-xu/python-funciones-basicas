# Función float():
permite convertir un número entero o una cadena numérica válida en un número de tipo decimal. Es útil cuando se necesitan realizar operaciones que requieren valores con decimales.
## Sintaxis:
float(valor)
## Ejemplo:
Un sistema de ventas recibe el precio de un producto como texto y necesita convertirlo en un número decimal para realizar cálculos.

```python
precio_texto = "25900"

precio_decimal = float(precio_texto)

print("Valor original:", precio_texto)
print("Valor convertido:", precio_decimal)
print("Tipo de dato:", type(precio_decimal))
```

## Explicación:
La función float() convierte la cadena "25900" en el número decimal 25900.0, permitiendo utilizarlo en operaciones matemáticas.
