# Función all()
Permite comprobar si todos los elementos de una colección cumplen una condición o tienen un valor verdadero. Devuelve True solamente cuando todos los elementos son verdaderos.
## Sintaxis: 
all(iterable)

```python
calificaciones = [3.5, 4.0, 3.8, 4.5]

todas_aprobadas = all(nota >= 3.0 for nota in calificaciones)

print("¿Todos los estudiantes aprobaron?", todas_aprobadas)
```

# Explicación: 
Verifica que todas las calificaciones sean iguales o superiores a 3.0.
