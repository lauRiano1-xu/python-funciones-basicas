# Función any()
Permite comprobar si al menos uno de los elementos de una colección tiene un valor verdadero. Devuelve True si encuentra al menos uno y False si ninguno es verdadero.
## Sintaxis: 
any(iterable)

```python
calificaciones = [2.5, 3.0, 4.2, 1.8]

hay_aprobados = any(nota >= 3.0 for nota in calificaciones)

print("¿Existe alguna nota aprobada?", hay_aprobados)
```

##  Explicación: 
Comprueba si existe al menos una calificación igual o superior a 3.0.
