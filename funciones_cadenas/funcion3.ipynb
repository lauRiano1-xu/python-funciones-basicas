# Función list()
Permite crear una lista a partir de un objeto iterable, como una cadena de texto, una tupla o un conjunto.
## Sintaxis:
list(iterable)

```python
ciudades = ("Bogotá", "Medellin", "Cali")
lista_ciudades = list(ciudades)
print(lista_ciudades)
```

## Explicación 
Convierte una tupla en una lista, permitiendo modificar posteriormente sus elementos.
