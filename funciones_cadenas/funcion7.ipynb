# Función enumerate()
Permite recorrer una colección obteniendo tanto el índice como el valor de cada elemento. Es especialmente útil cuando necesitamos conocer la posición de los elementos.
## Sintaxis: 
enummerate (iterable, start=0)

```python
lenguajes = ["Python", "Java", "JavaScript", "PHP"]

for posicion, lenguaje in enumerate(lenguajes, start=1):
    print(posicion, "-", lenguaje)
```

## Explicación: 
La función enumera los lenguajes de programación comenzando desde la posición 1.
