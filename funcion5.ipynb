# Función set():
Permite crear un conjunto de elementos. Una de sus principales características es que no almacena valores duplicados.
# Sintaxis: 
set(iterable)

```python
numeros = [1, 2, 3, 2, 4, 3, 5, 1]
numeros_unicos = set (numeros)
print(numeros_unicos)
```

## Explicación:
El conjunto elimina automáticamente los elementos repetidos. El orden de los elemetos de un conjunto no está garantizado.
